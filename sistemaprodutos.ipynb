```python
print("============SISTEMA DE ESTOQUE DE UMA PAPELARIA==============")

produtos=["papel ","lapis ","borracha ","apontador "]

precos = [3.45,4,2.50,10]

novo_produto = input("Digite o nome do novo produto: ")
novo_preco = float(input("Digite o valor do preco: "))

produtos.append(novo_produto)
precos.append(novo_preco)


print("Estoque atualizado")


for i,j in zip(produtos,precos):
  print(f"o produto {i}, custa preço:{j}")

if len(produtos) > 0:
    remover_produto= produtos.pop()
    remover_preco= precos.pop()
    print(f"O produto{remover_produto}, e o preço {remover_preco} foram removidos")
    print(f"{produtos},{precos}")

else:
    print("estoque esta vazio")
    print(f"{produtos},{precos}")




```

```python
#funcionamento while
#while -- condição:
#exemplo1
limite = int(input("Informe o numero limite: "))
inicio = 0

while inicio <= limite:
  print(inicio)
  inicio = inicio + 1

```

```python
#controle com while
senha_correta = "abc123"
tentativas = 0

while tentativas < 3:
  senha = input("Digite sua senha: ")
  if senha_correta == senha:
   print("Bem vindo ao Sistema ")
   break #quebra o while

  else:
    print("senha incorreta")
    tentativas = tentativas + 1
    print(f"numero de tentativas: {tentativas}")


```

```python
#atividade 1 (facil)
#fazer sistema que pergunta ao usuario se ele quer salvar
#uma informação em uma lista de carro

print("===========BEM VINDO A NOSSA CONCESSIONARIA GEN ==================")

carros = ["uno","fusca","corola"]
resposta = input("Deseja salvar os nomes dos carros disponiveis na loja? digite SIM para salvar: ")
if resposta.lower() == "sim":
  print(f"{carros}")


while True:
  novo_carro = input("quais carros deseja adicionar a lista? ou digite sair: ")
  carros.append(novo_carro)
  print("Carro inserido!!")

  if novo_carro.lower() == "sair":
    print("voce saiu do sistema!!")
    break
  else:
    for i in carros:
      print(f"{i}")




```
